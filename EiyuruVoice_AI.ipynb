```python
import zipfile
import os

def unzip_file(zip_path, extract_to):
    if os.path.exists(zip_path):
        print(f'{zip_path} çıkarılıyor...')
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_to)
        print(f'{zip_path} başarıyla {extract_to} dizinine çıkarıldı.')
    else:
        print(f'Hata: {zip_path} bulunamadı.')

# Dosyaları çıkar
unzip_file('/content/DiffSinger-master.zip', '/content/DiffSinger')
unzip_file('/content/OpenUtau-master.zip', '/content/OpenUtau')
unzip_file('/content/DiffSinger_欲音ルコ♀_V200.zip', '/content/Voicebank_Ruko')
```
